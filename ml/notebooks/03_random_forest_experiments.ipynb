Eksperimen training Random Forest Classifier, Pengaturan Jumlah Pohon, Validasi, Evaluasi Matriks, dan Analisis Korelasi Feature Importance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")
y_train = pd.read_csv("../data/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()

param_grid = {
    "n_estimators": [50, 100, 150],
    "max_depth": [4, 6, 8, None],
    "min_samples_split": [2, 5]
}

grid = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5, scoring="f1_weighted", n_jobs=-1)
grid.fit(X_train, y_train)
print("Parameter Terbaik:", grid.best_params_)
best_rf = grid.best_estimator_
```

```python
y_pred = best_rf.predict(X_test)
print("Akurasi:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Feature Importance
imp = pd.Series(best_rf.feature_importances_, index=X_train.columns).sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 5))
imp.plot(kind="barh", color="#10b981").invert_yaxis()
plt.title("Top 10 Feature Importance (Random Forest)")
plt.show()
```
